2. Write a NumPy program to create a 5x5 array with random values and find
the minimum and maximum values for each row.

```python
import numpy as np
empty_a = np.empty((5, 5), dtype=int)
print("Array:\n", empty_a)

for i in range(empty_a.shape[0]): 
    row_min = np.min(empty_a[i])
    row_max = np.max(empty_a[i])
    print(f"Row {i}: min = {row_min:}, max = {row_max:}")
```

3. Write a NumPy program to create a random vector of size 10 and sort it.

```python
import numpy as np
vec = np.random.randint(1,11, size=10)   
print(f"Random Vector:\n {vec}")

sorted_vec = np.sort(vec)
print(f"\nSorted vector:\n{sorted_vec}")

```

4. Write a NumPy program to find the most frequent value in an array.

```python
import numpy as np
a=np.array([1,6,3,4,8,8,3,8,9])
most_frequent = np.bincount(a).argmax()
print("The most frequent number is:", most_frequent)

```

5. Replace all even numbers in the 2D array with 0.

```python
import numpy as np

a = np.array([[1,2, 5], [3, 4, 6]])
print("Array:\n", a)
a[a% 2 == 0] = 0
print(f"\nArray after replacing even numbers with 0:\n", a)
```

7. Write a NumPy program to create a 3x3 array with random values and
subtract the mean of each row from each element.

```python
import numpy as np
a= np.empty((3, 3), dtype=int)
print("Random 3x3 array:\n", a)
row_means = a.mean(axis=1)
print("Row means:", row_means)
result=a - row_means
print("Resulting array after subtracting row means:\n", result) 
```

9. Write a NumPy program to create a 5x5 array with random values and find
the second-largest value in each column.

```python
import numpy as np

a = np.empty((5, 5), dtype=int)
print("Array:\n", a)


sorted_a = np.sort(a, axis=0)

second_largest = sorted_a[-2, :]

print("\nSecond largest value in each column:\n", second_largest)
```
